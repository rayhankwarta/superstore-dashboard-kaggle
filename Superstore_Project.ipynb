```python
import pandas as pd
import numpy as np

# Specifying the encoding to handle special characters in the Superstore dataset
df = pd.read_csv("/content/Sample - Superstore.csv", encoding='windows-1252')

# Correcting the invalid print.df() syntax
print(df)
```

```python
df.describe()
```

```python
df.info()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df['Shipping_Days'] = (df['Ship Date'] - df['Order Date']).dt.days
df['Order_Year'] = df['Order Date'].dt.year
df['Order_Month'] = df['Order Date'].dt.to_period('M').astype(str)
```

```python
df.info()
```

```python
df['Postal Code'] = df['Postal Code'].astype(str).str.zfill(5)
```

```python
cols_to_check = [col for col in df.columns if col != 'Row ID']
df_clean = df.drop_duplicates(subset=cols_to_check).copy()
```

```python
# Mengecek missing values di setiap kolom
print("=== Jumlah Missing Values Per Kolom ===")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0] if (missing_values > 0).any() else "Tidak ada missing values di kolom mana pun.\n")

# Mengecek duplikat data (secara keseluruhan)
print("=== Pengecekan Duplikat Data ===")
jumlah_duplikat = df.duplicated().sum()
print(f"Jumlah baris duplikat (identik sepenuhnya): {jumlah_duplikat}")

# Mengecek duplikat berdasarkan kolom penting (tanpa menyertakan 'Row ID')
cols_to_check = [col for col in df.columns if col != 'Row ID']
duplikat_tanpa_row_id = df.duplicated(subset=cols_to_check).sum()
print(f"Jumlah baris duplikat (tanpa menyertakan 'Row ID'): {duplikat_tanpa_row_id}")
```

```python
df_clean['Profit_Status'] = df_clean['Profit'].apply(lambda x: 'Loss' if x < 0 else 'Profit')
df_clean['Discount_Tier'] = pd.cut(
    df_clean['Discount'],
    bins=[-0.01, 0.0, 0.20, 0.50, 1.0],
    labels=['No Discount (0%)', 'Low (1-20%)', 'Medium (21-50%)', 'High (>50%)']
)
```

```python
df_clean.to_csv('superstore_cleaned.csv', index=False)
print(f"Data bersih siap! Jumlah baris akhir: {len(df_clean)}")
```
