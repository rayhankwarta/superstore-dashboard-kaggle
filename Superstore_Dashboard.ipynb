```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

# Visualisasi Data

```python
monthly_sales_profit = df[['Sales', 'Profit']].resample('M').sum()

plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(14, 7))

ax.plot(monthly_sales_profit.index, monthly_sales_profit['Sales'], color='b', label='Penjualan')
ax.plot(monthly_sales_profit.index, monthly_sales_profit['Profit'], color='g', label='Keuntungan')

ax.set_title('Tren Penjualan dan Keuntungan Bulanan', fontsize=16)
ax.set_xlabel('Bulan', fontsize=12)
ax.set_ylabel('Jumlah', fontsize=12)
ax.legend()
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig('sales_profit_trend.png')
```

Dari plot yang dihasilkan, Anda dapat melihat fluktuasi penjualan dan keuntungan dari bulan ke bulan. Biasanya, akan terlihat adanya puncak penjualan di akhir tahun (kuartal keempat), yang merupakan pola umum dalam bisnis ritel. Analisis ini membantu dalam perencanaan stok dan strategi pemasaran.

```python
df.reset_index(inplace=True)

category_sales = df.groupby(['Category', 'Sub-Category'])['Sales'].sum().reset_index()

plt.figure(figsize=(12, 6))
sns.barplot(data=category_sales.sort_values('Sales', ascending=False), x='Sales', y='Category', palette='viridis', ci=None)
plt.title('Total Penjualan per Kategori')
plt.xlabel('Total Penjualan')
plt.ylabel('Kategori')
plt.savefig('sales_by_category.png')
```

Visualisasi ini akan menunjukkan kategori dan sub-kategori produk yang menghasilkan penjualan tertinggi. Dari sini, Anda bisa melihat bahwa kategori seperti Technology dan Furniture biasanya menjadi pendorong utama penjualan, terutama produk seperti Phones dan Chairs. Ini adalah informasi penting untuk manajemen inventaris.

```python
plt.figure(figsize=(12, 8))
sns.barplot(data=category_sales.sort_values('Sales', ascending=False).head(10), x='Sales', y='Sub-Category', palette='plasma', ci=None)
plt.title('Top 10 Sub-Kategori Berdasarkan Penjualan')
plt.xlabel('Total Penjualan')
plt.ylabel('Sub-Kategori')
plt.savefig('top_10_subcategory_sales.png')
```

```python
region_sales = df.groupby('Region')['Sales'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=region_sales.sort_values('Sales', ascending=False), x='Region', y='Sales', palette='cividis')
plt.title('Total Penjualan per Wilayah')
plt.xlabel('Wilayah')
plt.ylabel('Total Penjualan')
plt.savefig('sales_by_region.png')
```

Chart ini akan menunjukkan perbandingan penjualan antar wilayah (Region). Anda dapat dengan cepat mengidentifikasi wilayah mana yang menjadi pasar terbesar. Informasi ini berguna untuk strategi ekspansi atau alokasi sumber daya pemasaran regional.

```python
plt.figure(figsize=(12, 7))
sns.scatterplot(data=df, x='Discount', y='Profit', hue='Category', palette='Set2', s=50)
plt.title('Hubungan antara Diskon dan Keuntungan')
plt.xlabel('Diskon')
plt.ylabel('Keuntungan')
plt.axhline(0, color='red', linestyle='--')
plt.legend(title='Kategori')
plt.savefig('discount_profit_relationship.png')
```

Scatter plot ini menunjukkan bahwa semakin besar diskon, keuntungan cenderung semakin kecil, bahkan bisa negatif (rugi). Terlihat jelas bahwa ketika diskon di atas 20%-30%, banyak transaksi yang merugi. Ini adalah wawasan krusial yang menunjukkan bahwa strategi diskon perlu dikelola dengan hati-hati agar tidak mengorbankan profitabilitas.
